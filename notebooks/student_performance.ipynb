# 1.0 Dataset Introduction

Low achievement and failure in core subjects such as mathematics are a long-standing concern in secondary education, as they reduce students' later educational and career opportunities. Grades are shaped not only by ability but also by a student's background and lifestyle, such as their parents' education, family relationships, study habits, social life and alcohol use.

This dataset records these factors alongside each student's grades across the school year, for students at two secondary schools in Portugal. It was created to test whether data mining could identify students at risk of poor results early enough for schools to step in with extra support. It supports both predicting a student's final grade (regression) and predicting whether they will pass or fail (classification).

## 1.1 Dataset Introduction

The dataset was retrieved from the UC Irvine Machine Learning Repository. It was collected by Paulo Cortez and Alice Silva (University of Minho) from school reports and questionnaires at two Portuguese secondary schools, and was introduced in Cortez and Silva (2008). It was donated to the repository in 2014 and is available under the Creative Commons Attribution 4.0 licence. This file is the mathematics course subset; the repository also holds a separate Portuguese language subset of 649 students. The dataset can be found at the following URL: 

https://archive.ics.uci.edu/dataset/320/student+performance

The dataset contains thirty-two (32) features, with one (1) additional target for a total of thirty-three (33) attributes over 395 rows. It has no missing values. The attributes are defined as the following types:

- <mark>Categorical (Binary)</mark>: school, sex, address, famsize, Pstatus, schoolsup, famsup, paid, activities, nursery, higher, internet, romantic Each takes one of two values, such as GP/MS, F/M, U/R or yes/no.
- <mark>Categorical (Nominal)</mark>: Mjob, Fjob, reason, guardian The values (e.g. teacher, health, services, at_home, other) have no natural order.
- <mark>Categorical (Ordinal)</mark>: Medu, Fedu (0 to 4); traveltime, studytime (1 to 4); famrel, freetime, goout, Dalc, Walc, health (1 to 5) These are stored as integers but are coded survey scales, for example 1 = very low to 5 = very high. The gaps between levels are not necessarily equal, so the values are ranked rather than truly numerical.
- <mark>Numerical (Discrete)</mark>: age (15 to 22), failures (0 to 3), absences (0 to 75 in this file), G1, G2 and G3 (each 0 to 20) All are whole numbers within a defined range.

There are no date or continuous attributes. G1 and G2 are first- and second-period grades, so they are strongly correlated with the final grade G3. Including them makes prediction considerably easier but much less useful in practice. Table 3 summarises the attributes.



| Attribute(s) | Data type | Variable type | Description |
|---|---|---|---|
| school, sex, address | String | Categorical (binary) | School attended, sex, urban/rural home |
| famsize, Pstatus | String | Categorical (binary) | Family size (≤3 / >3), parents living together or apart |
| schoolsup, famsup, paid, activities, nursery, higher, internet, romantic | String | Categorical (binary) | Yes/no: school support, family support, paid classes, activities, nursery, wants higher education, internet, relationship |
| Mjob, Fjob, reason, guardian | String | Categorical (nominal) | Parents' jobs, reason for choosing school, guardian |
| Medu, Fedu | Integer | Categorical (ordinal) | Parents' education level (0 none to 4 higher) |
| traveltime, studytime | Integer | Categorical (ordinal) | Banded travel and weekly study time (1 to 4) |
| famrel, freetime, goout, Dalc, Walc, health | Integer | Categorical (ordinal) | Family relations, free time, going out, weekday/weekend alcohol use, health (1 to 5) |
| age, failures, absences | Integer | Numerical (discrete) | Age, past class failures, school absences |
| G1, G2 | Integer | Numerical (discrete) | First- and second-period grades (0 to 20) |
| G3 | Integer | Numerical (discrete) | Target: final grade (0 to 20) |
